```python
# ติดตั้งและ import library
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# สร้างข้อมูลจำลอง
np.random.seed(42)
x = np.random.normal(50, 10, 100)
y = x * 0.8 + np.random.normal(0, 5, 100)   # Positive correlation
z = -x * 0.5 + np.random.normal(0, 5, 100)  # Negative correlation
w = np.random.normal(0, 10, 100)            # No correlation

# รวมเป็น DataFrame
df = pd.DataFrame({"X": x, "Y": y, "Z": z, "W": w})

# วาดกราฟ
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Positive
axes[0].scatter(df["X"], df["Y"], alpha=0.7)
axes[0].set_title("Positive Correlation (X vs Y)")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")

# Negative
axes[1].scatter(df["X"], df["Z"], alpha=0.7, color="red")
axes[1].set_title("Negative Correlation (X vs Z)")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Z")

# No correlation
axes[2].scatter(df["X"], df["W"], alpha=0.7, color="green")
axes[2].set_title("No Correlation (X vs W)")
axes[2].set_xlabel("X")
axes[2].set_ylabel("W")

plt.tight_layout()
plt.show()
df.corr()
```
